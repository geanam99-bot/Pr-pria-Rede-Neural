## 3. Normalização/Padronização dos dados

Nesta etapa os dados foram padronizados para preparar o dataset para o
treinamento da rede neural. A padronização é importante em modelos de
ANN pois evita que features com escalas diferentes influenciem de forma
desigual o processo de aprendizado.

### 3.1 Separação entre features e target

A coluna `target` foi separada das demais, já que representa a classe
a ser prevista e não deve ser normalizada.

```python
X = df.drop(columns=["target"])
y = df["target"]

X.head()
```

### 3.2 Aplicação do StandardScaler

Foi utilizado o `StandardScaler` da biblioteca `scikit-learn`, que
transforma cada feature para ter média 0 e desvio padrão 1. Essa técnica
é amplamente utilizada em redes neurais por facilitar a convergência
do algoritmo de treinamento.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

df_normalizado = pd.DataFrame(X_scaled, columns=X.columns)
df_normalizado["target"] = y.values

df_normalizado.head()
```

### 3.3 Verificação dos dados normalizados

Foi utilizado o método `describe()` para confirmar que as features
passaram a apresentar média próxima de 0 e desvio padrão próximo de 1.

```python
df_normalizado.describe()
```

### 3.4 Conclusão da normalização

As quatro features numéricas do dataset Iris foram padronizadas com
sucesso utilizando o `StandardScaler`, deixando todas as variáveis na
mesma escala. A coluna `target` foi mantida sem alterações, já que
representa a classe de cada registro.

```python
df_normalizado.to_csv(
    "/content/drive/MyDrive/Projeto_ANN/iris_normalizado.csv",
    index=False
)
```
